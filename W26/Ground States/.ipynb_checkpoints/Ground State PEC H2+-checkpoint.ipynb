# Ground State Potenntial Energy Curve for H2+ Using HF

```python
#makes a file containing all the energy for each bond length

# h2plus_hf_pec.py
import psi4
import numpy as np
import csv
import glob
import os

#choose basis (small to large):
#cc-pVDZ, cc-pVTZ, cc-pVQZ, cc-pV5Z,
#aug-cc-pVDZ, aug-cc-pVTZ, aug-cc-pVQZ, aug-cc-pV5Z"

basis = "aug-cc-pV5Z"

r_min, r_max, n_points = 0.2, 5.0, 1000          #max and min bond length, number of points
csv_out = "h2plus_hf_pec.csv"                    #name csv output file
psi4_out = "psi4_h2plus_hf.out"                  #name psi4 output file


psi4.set_output_file(psi4_out, False)            #(Names output file, overrides existing files)
psi4.set_memory("4 GB")                          #How much ram psi4 is allowed to use
psi4.set_num_threads(4)                          #number of threads to use

#set up psi4
psi4.set_options({
    "basis": basis,              #choose basis
    "reference": "rohf",         #Restricted Open-Shell Hartree-Fock, needed for open shell system, can also use uhf
    "e_convergence": 1e-10,      #convergence limits
    "d_convergence": 1e-10,
    "maxiter": 200,              #max iterations
})

#create an array of points
bond_lengths = np.linspace(r_min, r_max, n_points)

energies = []                    #empty list for energies

#run calculations
for R in bond_lengths:                    #runs over array
    #create molecule
    mol = psi4.geometry(f"""
        1 2                          #charge, multiplicity
        H                            #first hydrogen at (0,0,0)
        H 1 {R}                      #second hydrogen at distance R from hydrogen #1
        units angstrom               #units of R in angstroms
        symmetry c1                  #turns off symmetry
        """)

    psi4.core.clean()                     #removes scratch files, clears data
    E = psi4.energy("scf", molecule=mol)  #calculate energy using Self-Consistent Field (Hartree-Fock) for my mol
    energies.append((float(R), float(E))) #appends data point (R,E) to my list of energies
    #print(f"H2+  UHF/{basis}  R={R:7.4f} Å  E={E: .12f} Eh")

#write csv file
with open(csv_out, "w", newline="") as f:                 #opens csv file, overwrite existing file, prevent blank lines
    w = csv.writer(f)                                     #make csv writer
    w.writerow(["R_angstrom", f"E_ROHF_{basis}_hartree"]) #write a header row
    w.writerows(energies)                                 #adds my data to the file


#delete psi4 output files after run

extensions = ["*.out", "*.log", "*.clean"]
for ext in extensions:
    for file in glob.glob(ext):
        os.remove(file)

print(f"\nSaved {csv_out}  |  Psi4 output: {psi4_out}")   #print to show completion
```

# Ground State Potential Energy Curve for H2+ Using FIC

```python
#makes a file containing all the energy for each bond length

# h2plus_fci_pec.py
import psi4
import numpy as np
import csv
import glob
import os

#choose basis (small to large):
#cc-pVDZ, cc-pVTZ, cc-pVQZ, cc-pV5Z,
#aug-cc-pVDZ, aug-cc-pVTZ, aug-cc-pVQZ, aug-cc-pV5Z"

basis = "aug-cc-pV5Z"

r_min, r_max, n_points = 0.2, 5.0, 1000           #max and min bond length, number of points
csv_out = "h2plus_fci_pec.csv"                    #name csv output file
psi4_out = "psi4_h2plus_fci.out"                  #name psi4 output file

psi4.set_output_file(psi4_out, False)            #(Names output file, overrides existing files)
psi4.set_memory("4 GB")                          #How much ram psi4 is allowed to use
psi4.set_num_threads(4)                          #number of threads to use

#set up psi4
psi4.core.clean_options()        #resets options
psi4.set_options({
    "basis": basis,              #set basis
    "reference": "rohf",         #Restricted Open-Shell Hartree-Fock, needed for open shell system, can also use uhf
    "guess": "sad",              #initial guess
    "scf_type": "pk",            #compute exact integrals
    "e_convergence": 1e-10,      #convergence limits
    "d_convergence": 1e-10,
    "maxiter": 200,              #max iterations
})

#create an array of points
bond_lengths = np.linspace(r_min, r_max, n_points)

bond_lengths= bond_lengths[300:320]
for r in bond_lengths:
    print(r)
    
energies = []                    #empty list for energies

#run calculations
for R in bond_lengths:           #runs over array
    #create molecule
    
    mol = psi4.geometry(f"""
        1 2                          #charge, multiplicity
        H                            #first hydrogen at (0,0,0)
        H 1 {R}                      #second hydrogen at distance R from hydrogen #1
        units angstrom               #units of R in angstroms
        symmetry c1                  #turns off symmetry
        """)

    psi4.core.clean()                     #removes scratch files, clears data
    E = psi4.energy("fci", molecule=mol)  #calculate energy using Full Configuration Interaction for my mol
    energies.append((float(R), float(E))) #appends data point (R,E) to my list of energies
    print(E)

#write csv file
with open(csv_out, "w", newline="") as f:                 #opens csv file, overwrite existing file, prevent blank lines
    w = csv.writer(f)                                     #make csv writer
    w.writerow(["R_angstrom", f"E_FIC_{basis}_hartree"])  #write a header row
    w.writerows(energies)                                 #adds my data to the file

#delete psi4 output files after run
extensions = ["*.out", "*.log", "*.clean"]
for ext in extensions:
    for file in glob.glob(ext):
        os.remove(file)

print(f"\nSaved {csv_out}  |  Psi4 output: {psi4_out}")   #print to show completion
```
